```python
import pandas as pd
import gdown

files_id = {
    '1rgPe6OuyI-ZCl8j3yeK7SqNCbk67EidF': 'survey_results_public.csv',
    '1cJMs8RrvkKHHP-GzJdUkf_irdNW1lYIj': 'survey_results_schema.csv'
}

for id, name in files_id.items():
    url = f'https://drive.google.com/uc?id={id}'
    gdown.download(url, name)

results_df = pd.read_csv('survey_results_public.csv', low_memory=False)
schema_df = pd.read_csv('survey_results_schema.csv')
```

**Завдання 1.** Підрахунок загальної кількості респондентів

```python
total_respondents = results_df['ResponseId'].nunique()

print(f"Загальна кількість респондентів: {total_respondents}")
```

**Завдання 2.** Аналіз повноти відповідей респондентів

```python
# Визначаємо набір питань із schema_df та наявні колонки в results_df
schema_questions = set(schema_df['qname'].unique())
data_columns = set(results_df.columns)

# Перетин двох наборів питань
target_columns = list(schema_questions.intersection(data_columns))

count_respondents = results_df.dropna(subset=target_columns).shape[0]
print(f"Кількість респондентів, які відповіли на всі запитання: {count_respondents }")
```

**Завдання 3.** Статистичний аналіз досвіду респондентів

```python
workexp = results_df['WorkExp'].dropna()

# series для представлення результату
central_tendency = pd.Series({
    'Mean': workexp.mean(),
    'Median': workexp.median(),
    'Mode': workexp.mode()[0]
}).round(2)


print('Міри центральної тенденції для поля WorkExp:\n')
print(central_tendency.to_string())
```

**Завдання 4.** Аналіз віддаленої роботи

```python
remote_count = (results_df['RemoteWork'] == 'Remote').sum()

print(f'Кількість респондентів, які працюють віддалено: {remote_count}')
```

**Завдання 5.** Визначення популярності Python

```python
# Пошук та підрахунок користувачів Python
python_users = (
    results_df['LanguageHaveWorkedWith']
    .str.contains('python', case=False, na=False)
)
python_popularity = (python_users.sum() / total_respondents) * 100

print(f"Відсоток респондентів, які програмують на Python: {python_popularity:.1f}%")
```

**Завдання 6**. Аналіз шляхів навчання програмуванню

```python
# Підрахунок користувачів, які навчались онлайн
online_learners_count = (
    results_df['LearnCode']
    .str.contains('Online Courses or Certification', case=False, na=False)
    .sum()
)
print(f"Кількість респондентів, які навчалися через онлайн-курси: {online_learners_count}")
```

**Завдання 7.** Географічний аналіз компенсації Python-розробників

```python
# Групування за країною та агрегація для ConvertedCompYearly
geo_compensation = (
    results_df[python_users]
    .dropna(subset=['ConvertedCompYearly'])
    .groupby('Country')['ConvertedCompYearly']
    .agg(['mean', 'median'])
    .round(2)
)

geo_compensation = geo_compensation.sort_values(by='median', ascending=False).reset_index()

print('Компенсація Python-розробників за країнами:\n')
display(geo_compensation)
```

**Завдання 8.** Аналіз освіти найбільш оплачуваних спеціалістів

```python
top_compensation = (
    results_df[['ResponseId', 'ConvertedCompYearly', 'EdLevel']]
    .dropna()
    .sort_values(by='ConvertedCompYearly',ascending=False)
    .head(5)
    .reset_index(drop=True)
)

print("Рівні освіти респондентів з найвищою компенсацією:\n")
display(top_compensation)
```

**Завдання 9.** Аналіз популярності Python по віковим категоріям

```python
# Створюємо булеву колонку користувачів Python
results_df['is_python'] = python_users

# Групуємо за віковими категоріями
age_popularity = results_df.groupby('Age').agg(
    total_respondents=('ResponseId', 'count'),
    python_users=('is_python', 'sum')
).reset_index()

age_popularity['Python Percentage'] = (
    age_popularity['python_users'] / age_popularity['total_respondents'] * 100
).round(2)

final_table = age_popularity[['Age', 'Python Percentage']].copy()

print('Популярнісь Python за віком:\n')
display(final_table)
```

**Завдання 10.** Аналіз індустрій серед високооплачуваних віддалених працівників

```python
# Очищення даних
df = results_df.dropna(subset=[
    'ConvertedCompYearly', 'RemoteWork', 'Industry'
]).copy()

income_threshold = df['ConvertedCompYearly'].quantile(0.75)

# Фільтрація віддалених та високооплачуваних працівників
highly_paid_remote = df[
    (df['ConvertedCompYearly'] >= income_threshold) &
    (df['RemoteWork'] == 'Remote')
]

# Розподіл індустрій
industry_distribution = highly_paid_remote['Industry'].value_counts().reset_index(name='Count')

industry_distribution['Percentage'] = (
    industry_distribution['Count'] / industry_distribution['Count'].sum() * 100
).round(2)

print('Найпоширеніші індустрії серед високооплачуваних віддалених респондентів:\n')
display(industry_distribution)
```
